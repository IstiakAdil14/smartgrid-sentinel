## 4. Behavior Breakdown by Risk & Exception Condition

### B. Medium-Risk Alert Behavior (YELLOW Priority)
- **Trigger Condition:** Model predicts `Medium` risk (e.g., Transformer Load 68.5%, Demand 165 MW).
- **Mapped Priority:** `YELLOW (Warning Advisory)`.
- **Generated Explanation:** *"The grid is currently under moderate stress. The transformer is operating at 68.5% load, which indicates moderate loading."*
- **Action Protocol:** `• Avoid unnecessary high-power appliance usage during peak hours. • Charge essential devices.`

### C. High-Risk Alert Behavior (RED Priority)
- **Trigger Condition:** Model predicts `High` risk (e.g., Transformer Load 92.4%, Demand 245 MW).
- **Mapped Priority:** `RED (Urgent Emergency)`.
- **Generated Explanation:** *"The grid is showing signs of significant stress. The transformer is heavily loaded at 92.4%, so continued monitoring is important."*
- **Action Protocol:** `⚠ Charge essential devices immediately. ⚠ Limit non-essential electricity consumption.`
